```python
import yaml
from ultralytics import YOLO
```

```python
yaml_path = '/content/drive/MyDrive/easyPark_GraduationProject/OCR_dataset/data.yaml'

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

print(f"Number of classes: {data['nc']}")
print(f"\nClasses:")
for i, name in enumerate(data['names']):
    print(f"  {i}: {name}")
```

```python
model_ocr = YOLO("yolov8s.pt")
```

```python
model_ocr.train(
    data      = '/content/drive/MyDrive/easyPark_GraduationProject/OCR_dataset/data.yaml',
    epochs    = 50,
    batch     = 16,
    imgsz     = 640,
    cache     = True,
    augment   = True,
    patience  = 10,
    device    = 0,
    name      = 'ocr_character_detector',
    project   = '/content/drive/MyDrive/easyPark_GraduationProject/runs',
    save      = True,
    plots     = True,
)

print("YOLO #2 Training complete!")
```

```python
# Load the best trained YOLO #2 model
model_ocr = YOLO('/content/drive/MyDrive/easyPark_GraduationProject/runs/ocr_character_detector/weights/best.pt')
```

```python
# Run evaluation on test set
metrics = model_ocr.val(
    data   = '/content/drive/MyDrive/easyPark_GraduationProject/OCR_dataset/data.yaml',
    split  = 'test',
    device = 0,
    plots  = True,
)
```

```python
print("\nYOLO #2 Evaluation Results:")
print(f"  mAP@50       : {metrics.box.map50:.4f}")
print(f"  mAP@50-95    : {metrics.box.map:.4f}")
print(f"  Precision    : {metrics.box.mp:.4f}")
print(f"  Recall       : {metrics.box.mr:.4f}")
```
